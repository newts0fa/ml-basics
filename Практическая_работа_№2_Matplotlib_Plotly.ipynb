### Задание 1.1: Синусоида и косинусоида
Используя библиотеки NumPy и Matplotlib, постройте на одном графике функции синуса и косинуса на интервале от 0 до $2\pi$ Оформите график, добавив легенду и подписи к осям (`.legend()` `.ylabel()` `.xlabel()`)

### Задание 1.7: Интерактивный график линий

1. Используя библиотеку numpy, сгенерируйте одномерный массив, представляющий месяцы года (от 1 до 12).
2. Используя библиотеку numpy, сгенерируйте три одномерных массива, представляющих среднюю температуру воздуха в трех городах (например, Москва, Париж и Нью-Йорк) в соответствующие месяцы. Пусть температура варьируется в диапазоне от -10 до 30 градусов Цельсия.
3. Используя библиотеку plotly express, создайте интерактивный график линий с помощью функции [`plotly.express.line`](https://plotly.com/python-api-reference/generated/plotly.express.line.html).
4. В качестве параметров функции `line` передайте массив месяцев в качестве оси `x`, массивы температур в качестве оси `y`, а также названия городов в качестве параметра `color`, чтобы разделить линии по группам .
5. Настройте внешний вид графика, добавляя подписи к осям, заголовок графика и легенду [`.update_layout()`](https://plotly.com/python/creating-and-updating-figures/#updating-figure-layouts).

```python
import numpy as np
import pandas as pd
import plotly.express as px

# Создание массива месяцев через numpy
period = np.arange(1, 13)

# Данные по городам
city_a = [-10, -8, -2, 6, 12, 18, 22, 20, 14, 7, 0, -7] # Мск
city_b = [5, 7, 10, 13, 17, 20, 23, 22, 18, 14, 9, 6]   # Пар
city_c = [0, 2, 8, 15, 20, 25, 28, 27, 22, 16, 10, 4]  # НЙ

# Сборка данных в таблицу
weather_df = pd.DataFrame({
    "Месяц": np.tile(period, 3),
    "Градусы": city_a + city_b + city_c,
    "Локация": ["Москва"] * 12 + ["Париж"] * 12 + ["Нью-Йорк"] * 12
})

# Отрисовка
dynamic_chart = px.line(weather_df,
                         x="Месяц",
                         y="Градусы",
                         color="Локация",
                         title="Климатические изменения по городам")

# Финальная настройка лейаута
dynamic_chart.update_layout(
    xaxis_title="Порядковый номер месяца",
    yaxis_title="Температура (C°)",
    legend_title="Города"
)

dynamic_chart.show()
```

### Задание 1.8: Интерактивный график столбцов

**Цель:** Визуализировать интерактивный график столбцов, отображающий распределение населения по странам мира.

1. Используя библиотеку numpy, сгенерируйте одномерный массив, представляющий названия 10 стран мира (например, Россия, Китай, США и т.д.).
2. Используя библиотеку numpy, сгенерируйте одномерный массив, представляющий население каждой из выбранных стран в миллионах человек. Пусть население варьируется в диапазоне от 10 до 1500 миллионов человек.
3. Используя библиотеку plotly express, создайте интерактивный график столбцов с помощью функции [`plotly.express.bar`](https://plotly.com/python-api-reference/generated/plotly.express.bar.html?highlight=plotly%20express%20bar#plotly.express.bar).
4. В качестве параметров функции `bar` передайте массив названий стран в качестве оси `x`, массив населения в качестве оси `y`, а также массив населения в качестве параметра `color`, чтобы изменять цвет столбцов в зависимости от их значения.
5. Настройте внешний вид графика, добавляя подписи к осям, заголовок графика и цветовую шкалу [`.update_layout()`](https://plotly.com/python/creating-and-updating-figures/#updating-figure-layouts).

```python
import numpy as np
import plotly.express as px

# Генерация массива названий 10 стран
countries = np.array([
    'Китай', 'Индия', 'США', 'Индонезия', 'Пакистан',
    'Нигерия', 'Бразилия', 'Бангладеш', 'Россия', 'Мексика'
])

# Установка seed для воспроизводимости результатов
np.random.seed(42)

# Генерация массива населения (в миллионах человек)
# Используем равномерное распределение для разнообразия значений
population = np.random.uniform(10, 1500, size=10)
# Округляем до целых чисел для лучшей читаемости
population = np.round(population).astype(int)

print("Страны:", countries)
print("Население (млн чел):", population)

# Создание интерактивного столбчатого графика
fig = px.bar(
    x=countries,  # Названия стран на оси X
    y=population,  # Население на оси Y
    color=population,  # Цвет столбцов зависит от населения
    color_continuous_scale='Viridis',  # Цветовая шкала
    title='Население стран мира (в миллионах человек)',
    labels={'x': 'Страна', 'y': 'Население (млн чел)', 'color': 'Население'}
)

# Настройка внешнего вида графика
fig.update_layout(
    xaxis_title='Страны',
    yaxis_title='Население (миллионы человек)',
    title_font_size=20,
    xaxis_tickangle=-45,  # Поворот подписей стран для лучшей читаемости
    plot_bgcolor='lightgray',  # Цвет фона графика
    width=900,  # Ширина графика
    height=500  # Высота графика
)

# Настройка цветовой шкалы
fig.update_coloraxes(colorbar_title='Население<br>(млн чел)')

# Отображение графика
fig.show()

# Сохранение графика в HTML файл (опционально)
# fig.write_html('population_chart.html')
```

4. Для каждого подграфика укажите заголовок и метки осей.

6. Если графики накладываются друг на друга, то для добавления расстояния между ними, используйте метод [`.subplots_adjust()`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.subplots_adjust.html)

#### Описание модели SIR:

Модель SIR (Susceptible, Infected, Recovered) представляет собой классическую эпидемиологическую модель, которая разделяет популяцию на три основные группы:
- **S (Susceptible)**: Часть популяции, восприимчивая к инфекции, но еще не зараженная.
- **I (Infected)**: Часть популяции, зараженная инфекцией и способная передавать ее другим.
- **R (Recovered)**: Часть популяции, выздоровевшая после инфекции и получившая иммунитет.

#### Математический аппарат модели SIR:

1. **Дифференциальные уравнения модели SIR:**

   - $\frac{dS}{dt} = -\beta \frac{SI}{N}$ — уменьшение числа восприимчивых из-за заражения.
   - $\frac{dI}{dt} = \beta \frac{SI}{N} - \gamma I$ — изменение числа инфицированных, учитывая новые случаи заражения и выздоровление.
   - $\frac{dR}{dt} = \gamma I$ — увеличение числа выздоровевших благодаря выздоровлению инфицированных.

   Где:
   - $N$ — общая численность популяции (постоянная величина).
   - $\beta$ — коэффициент заразности, определяющий вероятность передачи инфекции от инфицированных к восприимчивым.
   - $\gamma$ — коэффициент выздоровления, обратный среднему периоду инфекционности.

  - В этом классе происходит моделирование взаимодействия между хищниками и жертвами. Используя метод `run`, модель выполняет итерации, на каждом шаге рассчитывая изменения в популяциях с помощью уравнений, описанных выше, и обновляя их значения.


Ваша реализация:

```python
# Импортируем необходимые библиотеки
import numpy as np
import plotly.express as px

# Создаем класс для моделирования популяции жертв
class Prey:
    def __init__(self, x0, alpha):
        # Инициализируем начальное количество жертв и коэффициент роста
        self.x = x0
        self.alpha = alpha

    def grow(self):
        # Рассчитываем прирост популяции жертв без учета хищников
        return self.alpha*self.x

    def die(self, y, beta):
        # Рассчитываем убыль популяции жертв из-за хищников
        return -beta*self.x*y

    def update(self, dx):
        # Обновляем количество жертв с учетом прироста и убыли
        self.x += dx

# Создаем класс для моделирования популяции хищников
class Predator:
    def __init__(self, y0, gamma):
        # Инициализируем начальное количество хищников и коэффициент смертности
        self.y = y0
        self.gamma = gamma

    def grow(self, x, delta):
        # Рассчитываем прирост популяции хищников за счет питания жертвами
        return delta*x*self.y

    def die(self):
        # Рассчитываем убыль популяции хищников из-за естественной смертности
        return -self.gamma*self.y

    def update(self, dy):
        # Обновляем количество хищников с учетом прироста и убыли
        self.y += dy

# Создаем класс для моделирования динамики популяции хищник-жертва
class EcoSystem:
    def __init__(self, x0, y0, alpha, beta, gamma, delta, t_max, dt):
        # Инициализируем параметры модели
        self.prey = Prey(x0, alpha)
        self.predator = Predator(y0, gamma)

        self.beta = beta
        self.delta = delta
        self.t = 0
        self.t_max = t_max
        self.dt = dt

        self.x_list = [x0]
        self.y_list = [y0]
        self.t_list = [0]

    def run(self):
        # Запускаем моделирование
        while self.t < self.t_max:
            # Рассчитываем изменения популяций за один шаг по времени
            dx, dy = 0, 0
            dx += self.prey.grow()
            dy += self.predator.grow(self.prey.x, self.delta)
            dx += self.prey.die(self.predator.y,self.beta)
            dy += self.predator.die()
            dx *= self.dt # домножаем на dt, исходя из формулы dx/dt = ...
            dy *= self.dt
            # Обновляем популяции с учетом изменений
            self.prey.update(dx)
            self.predator.update(dy)
            # Обновляем время
            self.t += self.dt
            # Добавляем результаты в списки
            self.x_list.append(self.prey.x)
            self.y_list.append(self.predator.y)
            self.t_list.append(self.t)

    def plot(self):
      # Преобразуем списки в массивы NumPy
      t = np.array(self.t_list)
      stats = {
          "Жертвы": np.array(self.x_list),
          "Хищники": np.array(self.y_list)
          }
      x = np.array(self.x_list)
      y = np.array(self.y_list)
      # Создаем линейный график изменения популяций во времени
      fig_line = px.line(stats, x=t, y=stats, title='Динамика популяции хищник-жертва во времени')
      fig_line.update_layout(xaxis_title="Время, единицы",
                             yaxis_title="Количество особей",
                             legend_title_text="Виды популяций")
      # Создаем фазовый портрет
      fig_phase = px.scatter(x=x, y=y, title='Фазовый портрет системы хищник-жертва',
                             labels={"x":"Популяция жертв", "y":"Популяция хищников"})
      # Отображаем оба графика
      fig_line.show()
      fig_phase.show()
```

```python
# Создаем объект экосистемы с заданными параметрами
ecosystem = EcoSystem(x0=10, y0=5, alpha=0.1, beta=0.01, gamma=0.1, delta=0.01, t_max=1000, dt=0.1)
# Запускаем моделирование
ecosystem.run()
# Визуализируем результаты
ecosystem.plot()
```
